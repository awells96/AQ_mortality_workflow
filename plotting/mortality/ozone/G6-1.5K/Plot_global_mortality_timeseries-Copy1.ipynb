```python
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from utils.utils import autosize_figure
```

```python
# === Path config ===
DIR = "/glade/work/awells/air_quality/CESM2/mortality/ozone/"

ds = []
for i in range(1, 4):
    ds.append(xr.open_dataset(f"/glade/work/awells/air_quality/CESM2/mortality/ozone/Global_mortality_stats_CESM2_SSP245_G6_0{i}_2020_2083.nc"))
ds_all = xr.concat(ds, "ensemble")
```

```python
da_mean = ds_all["M_mean"]
da_median = ds_all["M_median"]
da_lower95 = ds_all["M_lower95"]
da_upper95 = ds_all["M_upper95"]
```

```python
fig = plt.figure(figsize=autosize_figure(1, 1, xscale_factor=1.2))
gs = GridSpec(1, 1) # rows, columns

ax1 = fig.add_subplot(gs[0, 0])
da_median.mean("ensemble").plot(color='tab:purple', label="median")
da_mean.mean("ensemble").plot(color='grey', label="mean")
#ax1.fill_between(da_mean.year, da_lower95.mean("ensemble"), da_upper95.mean("ensemble"), alpha=0.25, label="95% CI")
plt.title("Global mortality\n SSP245_G6")
plt.legend()
"""
ax2 = fig.add_subplot(gs[1, 0])
ssp245_time = ssp245.sel(year=slice(2035, 2083))
for ens_num in range(1, 3):
    (g6 - ssp245_time).sel(ensemble=ens_num, quantile="mean").plot(color='k', alpha=0.4, linewidth=0.5)
(g6.mean("ensemble") - ssp245_time.mean("ensemble")).sel(quantile="mean").plot(color='k')
plt.axhline(0, color='k', linestyle='--')
plt.title("Difference in Global Mortality")"""
```
