# Plotting the ensemble range of temperature which is split between hot/cold mortality

Using n=10 to assess the effects of internal varaiability

```python
import os
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import Normalize
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from utils.utils import get_scenario_config, standardise_latlon
from utils.utils import land_filter, adjust_longitude
import config
from utils.utils import require_dir
import pathlib
```

```python
def split_temperature(temp, tmrel):
    """ Split temperature to hot and cold, like we do with mortality """
    print("splitting temp")
    # Create mask for cold related and heat related deaths
    cold = xr.where(temp < tmrel, 1, 0)
    hot = xr.where(temp > tmrel, 1, 0)
    # select mortality, nan where cold/hot is 0
    cold_death_temps = temp.where(cold == 1).drop_vars("annual_temperature")
    heat_death_temps = temp.where(hot == 1).drop_vars("annual_temperature")

    cold_temps = cold_death_temps.mean("time")
    hot_temps = heat_death_temps.mean("time")

    return cold_temps, hot_temps
```

```python
def process_temperature(model, scenario, year, ensemble_members):
    # === Path config ===
    TEMP_DIR = "/glade/derecho/scratch/arielm/Olympics/ARISE-SAI-1.5-downscaled/TREFHT/SSP2-4.5/"
    TMREL_DIR = require_dir(pathlib.Path(config.WORK_ROOT) / "GBD23" / "TMREL" / "summaries")
    # Load mean tmrel at all gridcells
    tmrel_map = xr.open_dataarray(TMREL_DIR/"tmrel_mean_0.25_map.nc")

    low_ensembles = []
    high_ensembles = []
    for ens_num in ensemble_members:
        print(year, f"{ens_num:02d}")
        file = f"bc_TREFHT_{scenario[:-3]}_ens0{ens_num:02d}_{year}.nc"
        file_path = os.path.join(TEMP_DIR, file)
        da = adjust_longitude(standardise_latlon(xr.open_dataarray(file_path) - 273.15))
        annual = da.mean("time")        # (lat, lon), assumes one year per file
        # Find tmrel based on annual mean temperature at each grid cell
        tmrel_everywhere = tmrel_map.sel(annual_temperature=annual, method="nearest")
        del annual
        low_temp, high_temp = split_temperature(da, tmrel_everywhere)
        del tmrel_everywhere, da
        low_ensembles.append(low_temp)
        high_ensembles.append(high_temp)
    low_temp_ens = xr.concat(low_ensembles, dim=xr.DataArray(np.arange(1, len(low_ensembles)+1), dims="ensemble", name="ensemble"))
    high_temp_ens = xr.concat(high_ensembles, dim=xr.DataArray(np.arange(1, len(high_ensembles)+1), dims="ensemble", name="ensemble"))


    return low_temp_ens, high_temp_ens
```

```python
def plot_mortality_ensemble_grid(
    da, model, scenario, year, temp,
    SAVE_DIR=None, ncols=5, anomaly=False, robust_pct=99, vmax=None,
    member_dim="n", lat_name="lat", lon_name="lon",
    cmap="RdBu_r", borders=False, panel_width=3.2,
):
    """
    da: DataArray with dims (lat, lon, n) in any order.
    anomaly=True plots each member minus the ensemble mean (shows internal variability).
    vmax: symmetric colour limit; if None, uses the `robust_pct` percentile of |values|.
    """
    n = da.sizes[member_dim]
    nrows = int(np.ceil(n / ncols))

    plot_da = da - da.mean(member_dim) if anomaly else da

    # One shared, zero-centred scale for ALL panels (essential for comparing members)
    if vmax is None:
        vmax = float(np.nanpercentile(np.abs(plot_da.values), robust_pct))
    vmin = -vmax if anomaly else 0
    norm = Normalize(vmin=vmin, vmax=vmax)

    projection, crs = ccrs.Robinson(), ccrs.PlateCarree()
    fig, axes = plt.subplots(
        nrows, ncols,
        figsize=(panel_width * ncols, panel_width * 0.55 * nrows + 1.2),
        subplot_kw={"projection": projection},
        constrained_layout=True,
    )
    axes = np.atleast_1d(axes).ravel()

    for i in range(n):
        ax = axes[i]
        member = land_filter(plot_da.isel({member_dim: i})).transpose(lat_name, lon_name)
        im = ax.pcolormesh(
            member[lon_name], member[lat_name], member.values,
            transform=crs, cmap=cmap, norm=norm, shading="auto", rasterized=True,
        )
        ax.coastlines(resolution="110m", linewidth=0.4)
        if borders:
            ax.add_feature(cfeature.BORDERS.with_scale("110m"), linewidth=0.2)
        ax.set_title(f"Member {i + 1}", fontsize=8)

    for ax in axes[n:]:          # hide unused panels
        ax.set_visible(False)

    cbar = fig.colorbar(im, ax=axes[:n].tolist(), orientation="horizontal",
                        shrink=0.5, aspect=40, pad=0.02, extend="both")
    label = "Anomaly from ensemble mean in" if anomaly else "Total"
    cbar.set_label(f"{label} annual mean {temp} surface temperature", fontsize=11)

    fig.suptitle(f"Surface temperature, {model} {scenario} {year} ({n} members)", fontsize=13)

    if SAVE_DIR is not None:
        suffix = "anomaly" if anomaly else "raw"
        out_file = f"{temp}_surface_temperature_{model}_{scenario}_{year}_ens{n}_{suffix}.png"
        fig.savefig(os.path.join(SAVE_DIR, out_file), dpi=300)
    return fig, axes
```

```python
# === Scenario and path config ===
# Set to whatever scenario and model you want
# Function returns error if not recognised
SAVE_DIR = require_dir(pathlib.Path(config.PLOTTING_ROOT) / "temperature")

model = "CESM2"
scenario = "SSP245_LE"

configs = get_scenario_config(model, scenario)
ensemble_members = configs["ensemble_members"]

plotting_year = 2050

# Calculate yearly mortality per 100k
low_temp, high_temp = process_temperature(model, scenario, plotting_year, ensemble_members)
```

```python
plot_mortality_ensemble_grid(low_temp, model, scenario, plotting_year, "low",
                             member_dim="ensemble", cmap="viridis", borders=True,
                             SAVE_DIR=SAVE_DIR)
```

```python
plot_mortality_ensemble_grid(high_temp, model, scenario, plotting_year, "high",
                             member_dim="ensemble", cmap="viridis", borders=True,
                             SAVE_DIR=SAVE_DIR)
```

```python
plot_mortality_ensemble_grid(low_temp, model, scenario, plotting_year, "low",
                             anomaly=True, member_dim="ensemble", borders=True,
                             SAVE_DIR=SAVE_DIR)
```

```python
plot_mortality_ensemble_grid(high_temp, model, scenario, plotting_year, "high",
                             anomaly=True, member_dim="ensemble", borders=True,
                             SAVE_DIR=SAVE_DIR)
```
