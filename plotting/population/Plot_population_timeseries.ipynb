# Plot population time series

```python
import os
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from utils.utils import autosize_figure
```

```python
# === Path config ===
SAVE_DIR = "/glade/u/home/awells/air_quality_project/plotting/population/"

ssp = "ssp2"
dates = "2000-2100"

POP_DIR = f"/glade/work/awells/air_quality/SSP_pop/{ssp.upper()}/"
pop_file = f"{ssp}_coarse_grid_annual_{dates}.nc"
pop_path = os.path.join(POP_DIR, pop_file)
pop = xr.open_dataarray(pop_path)

pop_timeseries = pop.sum(dim=("lat", "lon"))
```

```python
fig = plt.figure(figsize=autosize_figure(1, 1, xscale_factor=1.2))
gs = GridSpec(1, 1)  # rows, columns

# First panel
ax = fig.add_subplot(gs[0, 0], frameon=True)
cb = pop_timeseries.plot(linewidth=2)

plt.rcParams.update({'font.size': 16})

# Grid styling
ax.grid(True, linewidth=0.7, alpha=0.4)

# Title & labels
ax.set_title(f"{ssp.upper()} Population Projection")
ax.set_xlabel("Year")
ax.set_ylabel("Population (billions)")
ax.tick_params(axis="both")

ax = plt.gca()
for spine in ax.spines.values():
    spine.set_visible(False)

ax.tick_params(axis="x", length=0)
ax.tick_params(axis="y", length=0)

plt.tight_layout()

out_file = f"{ssp.upper()}_population_projection_timeseries.png"
out_path = os.path.join(SAVE_DIR, out_file)
plt.savefig(out_path)
```
