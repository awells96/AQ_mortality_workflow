```python
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from utils.utils import autosize_figure
```

```python
# === Path config ===
POP_DIR = "/glade/work/awells/air_quality/SSP_pop/SSP2/"

pop_ssp2 = xr.open_dataarray(f"{POP_DIR}ssp2_total_regrid_annual_2000-2100.nc")
```

```python
pop_timeseries = pop_ssp2.sum(dim=("lat", "lon"))
```

```python
fig = plt.figure(figsize=autosize_figure(1, 1, xscale_factor=1.5))
gs = GridSpec(1, 1) # rows, columns

# First panel
ax = fig.add_subplot(gs[0, 0], frameon=True)
cb = pop_timeseries.plot()
plt.grid()
plt.title("SSP2 Population", fontweight='bold')

plt.tight_layout()
```
