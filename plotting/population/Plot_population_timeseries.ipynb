```python
import os
import xarray as xr
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
from utils.utils import autosize_figure
```

```python
# === Path config ===
POP_DIR = "/glade/work/awells/air_quality/SSP_pop/SSP2/"
SAVE_DIR = "/glade/u/home/awells/air_quality_project/plotting/population/"

pop_file = "ssp2_coarse_grid_annual_2000-2100.nc"
pop_path = os.path.join(POP_DIR, pop_file)
pop = xr.open_dataarray(pop_path)

pop_timeseries = pop.sum(dim=("lat", "lon"))
```

```python
fig = plt.figure(figsize=autosize_figure(1, 1, xscale_factor=1.2))
gs = GridSpec(1, 1)  # rows, columns

# First panel
ax = fig.add_subplot(gs[0, 0], frameon=True)
cb = pop_timeseries.plot(linewidth=2)

# Grid styling
ax.grid(True, linestyle="--", linewidth=0.7, alpha=0.6)

# Title & labels
ax.set_title("SSP2 Population", fontsize=12, fontweight="bold")
ax.set_xlabel("Year", fontsize=12)
ax.set_ylabel("Population (millions)", fontsize=12)
ax.tick_params(axis="both", labelsize=10)

plt.tight_layout()

out_file = "SSP2_population_projection_timeseries.png"
out_path = os.path.join(SAVE_DIR, out_file)
plt.savefig(out_path)
```
