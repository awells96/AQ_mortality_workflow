```python
import os
import xarray as xr
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.gridspec import GridSpec
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from utils.utils import autosize_figure, land_filter, adjust_longitude
from utils.utils import get_scenario_config
```

```python
def load_MDA8_ensemble_members(model, scenario, config):
    # === Path config ===
    MDA8_DIR = f"/glade/work/awells/air_quality/{model}/ozone/MDA8/"

    ensemble_members = config["ensemble_members"]
    years = config["years"]
    dates = f"{years.start}0101-{years.stop}1231"

    ensemble = []
    for ens_num in ensemble_members:
        print(f"Processing ensemble number {ens_num:02d}")
        file = f"MDA8_{model}_{scenario}_{ens_num:02d}_{dates}.nc"
        path = os.path.join(MDA8_DIR, file)
        da = xr.open_dataarray(path)
        ensemble.append(da)

    new_da = xr.concat(
        ensemble,
        dim=xr.DataArray(
            np.arange(1, len(ensemble)+1),
            dims="ensemble",
            name="ensemble"))

    return new_da
```

```python
def process_mda8_for_plotting(model, scenario, config):
    # Load ensemble members
    da = load_MDA8_ensemble_members(model, scenario, config)
    
    # Calculate end of scenario mean (10 years)
    end_idx = da.sizes["time"]
    # Select the last 120 months
    da_last10 = da.isel(time=slice(end_idx - 120, end_idx))
    
    # Get the first and last time values
    start_time = da_last10.time[0].item()
    end_time = da_last10.time[-1].item()
    
    # Extract the years
    first_year = start_time.year
    last_year = end_time.year
    
    # Calculate ensemble and temporal mean
    da_mean = da_last10.mean(dim=("ensemble", "time"))
    return da_mean, first_year, last_year
```

```python
def plot_mda8(da, model, scenario, start_year, end_year, SAVE_DIR, land=False):
    # Create figure and layout
    fig = plt.figure(figsize=autosize_figure(1, 1))
    gs = GridSpec(2, 1, height_ratios=[15, 1])

    projection = ccrs.Robinson()
    crs = ccrs.PlateCarree()

    country_borders = cfeature.NaturalEarthFeature(
        category='cultural',
        name='admin_0_boundary_lines_land',
        scale='50m',
        facecolor='none')

    if land is True:
        da = land_filter(da)

    # Plot map
    ax = fig.add_subplot(gs[0, 0], projection=projection)
    cb = da.plot(
        transform=crs,
        add_colorbar=False,
        subplot_kws={'projection': projection}
    )
    ax.coastlines(resolution="50m", linewidth=0.75)
    ax.add_feature(country_borders, edgecolor='k', linewidth=0.75)
    plt.title(f"MDA8: 8hr Daily Maximum\n {model} {scenario} {start_year}-{end_year}", fontsize=16)

    # Colorbar
    cax = fig.add_subplot(gs[1, 0])
    col_bar = plt.colorbar(cb, cax=cax, orientation='horizontal')
    col_bar.set_label("Ozone (ppb)", fontsize=13)

    plt.tight_layout()

    if land is True:
        out_file = f"MDA8_land_{model}_{scenario}_{start_year}-{end_year}.png"
    else:
        out_file = f"MDA8_{model}_{scenario}_{start_year}-{end_year}.png"
    out_path = os.path.join(SAVE_DIR, out_file)
    plt.savefig(out_path)
    return
```

```python
# === Scenario and path config ===
# Set to whatever scenario and model you want
# Function returns error if not recognised
SAVE_DIR = "/glade/u/home/awells/air_quality_project/plotting/ozone/MDA8/"

model = "CESM2"
scenario = "G6-1.5K"

config = get_scenario_config(model, scenario)

da, first_year, last_year = process_mda8_for_plotting(model, scenario, config)

# Plotting without ocean
plot_mda8(da, model, scenario, first_year, last_year, SAVE_DIR, land=True)
# Plotting with ocean
plot_mda8(da, model, scenario, first_year, last_year, SAVE_DIR, land=False)
```

```python
# === Plotting differences between scenarios ===
# Set to whatever model you want and the two scenarios you want to compare
# Function returns error if not recognised

SAVE_DIR = "/glade/u/home/awells/air_quality_project/plotting/ozone/MDA8/"

model = "CESM2"
scenario = ["G6-1.5K", "SSP245_G6"]
scenario_name = f"{scenario[0]} - {scenario[1]}"

config1 = get_scenario_config(model, scenario[0])
config2 = get_scenario_config(model, scenario[1])

da1, first_year1, last_year1 = process_mda8_for_plotting(model, scenario[0], config1)
da2, first_year2, last_year2 = process_mda8_for_plotting(model, scenario[1], config2)

da = da1 - da2  # Check this is the way round you want it

# Plotting without ocean
plot_mda8(da, model, scenario_name, first_year1, last_year1, SAVE_DIR, land=True)
# Plotting with ocean
plot_mda8(da, model, scenario_name, first_year2, last_year2, SAVE_DIR, land=False)
```
