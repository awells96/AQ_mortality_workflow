# Test the bias corrected and downscaled temperature data for anomalies

```python
import os
import xarray as xr
```

### Bias corrected and downscaled 2015 SSP data from Ariel

```python
TEMP_DIR = "/glade/derecho/scratch/arielm/Olympics/ARISE-SAI-1.5-downscaled/TREFHT/SSP2-4.5"
file = "bc_TREFHT_SSP245_ens001_2015.nc"
file_path = os.path.join(TEMP_DIR, file)
temp = xr.open_dataarray(file_path)

print("temp range (K): ", float(temp.min().round(2)), float(temp.max().round(2)))
temp_C = temp - 273.15
print("temp range (C): ", float(temp_C.min().round(2)), float(temp_C.max().round(2)))
annual = temp_C.mean("time")
print("annual mean range (C): ", float(annual.min().round(2)), float(annual.max().round(2)))
```
