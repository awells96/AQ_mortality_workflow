```python
import xarray as xr
import regionmask
import numpy as np
```

```python
# === Latitude weighting mean ===
def weighted_mean(da):
    weights = np.cos(np.deg2rad(da.lat))
    weights.name = "weights"
    new_da = da.weighted(weights).mean(("lon", "lat"))
    return new_da


# === Removing ocean ===
def land_filter(da):
    # Load in land fraction dataset
    land_110 = regionmask.defined_regions.natural_earth_v4_1_0.land_110
    da = da.where(land_110.mask_3D(da).squeeze())
    return da


# === Concatenate ensemble members ===
def merge_global_mean_ensembles(VAR_DIR, VAR, scenario, dates, land_mask=False):
    # land_mask=True allows for the land filter function to remove the ocean
    ens = []

    for ens_num in range(1, 11):
        print(f"Processing ensemble member {ens_num:02d}")
        if scenario == "ARISE":
            year_range = range(2035, 2069)
        elif scenario == "SSP245":
            year_range = range(2020, 2069)

        years = []
        for year in year_range:
            da = xr.open_dataarray(f"{VAR_DIR}{VAR}_CESM2_{scenario}_{ens_num:02d}_{dates}.nc").sel(year=year)

            if land_mask is True:
                print("Filtering ocean")
                da = land_filter(da)

            years.append(weighted_mean(da))  # Latitude weighted mean

        ens.append(xr.concat(years, "time"))

    new_da = xr.concat(ens, dim=xr.DataArray(np.arange(1, len(ens)+1), dims="ensemble", name="ensemble"))

    return new_da
```

```python
# === Path config ===
OSDMA8_DIR = "/glade/work/awells/air_quality/CESM/ozone/OSDMA8_BC/"
VAR = "OSDMA8_BC"
```

```python
# === ARISE ===
SCENARIO = "ARISE"
DATES = "2035-2068"

arise_land = merge_global_mean_ensembles(OSDMA8_DIR, VAR, SCENARIO, DATES, land_mask=True)

arise_land.to_netcdf(f"{OSDMA8_DIR}{VAR}_globalmean_CESM2_{SCENARIO}_{DATES}.nc")
```

```python
# === SSP245 ===
SCENARIO = "SSP245"
DATES = "2020-2068"

ssp245_land = merge_global_mean_ensembles(OSDMA8_DIR, VAR, SCENARIO, DATES, land_mask=True)

ssp245_land.to_netcdf(f"{OSDMA8_DIR}{VAR}_globalmean_CESM2_{SCENARIO}_{DATES}.nc")
```
