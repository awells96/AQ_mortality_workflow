```python
import xarray as xr
import numpy as np
```

```python
# === Latitude weighting sum ===
def weighted_sum(da):
    weights = np.cos(np.deg2rad(da.lat))
    weights.name = "weights"
    new_da = da.weighted(weights).sum(("lon", "lat"))
    return new_da
```

```python
# === Path config ===
MORTALITY_DIR = "/glade/work/awells/air_quality/CESM/mortality/"
SCENARIOS = ["ARISE", "SSP245"]

# === Main loop ===
for scenario in SCENARIOS:
    ensembles = []
    for ens_num in range(1, 11):
        print(f"Processing {scenario}, Ensemble {ens_num:02d}")
        if scenario == "ARISE":
            dates = "2035-2068"
        elif scenario == "SSP245":
            dates = "2020-2068"

        da = xr.open_dataarray(f"{MORTALITY_DIR}Mortality_CESM2_{scenario}_{ens_num:02d}_{dates}.nc")

        global_sum = weighted_sum(da)
        ensembles.append(global_sum)

    ensemble_sum = xr.concat(ensembles, dim=(xr.DataArray(range(1, 11), dims="ensemble", name="ensemble")))

    print(f"Saving global sum in {MORTALITY_DIR}")
    ensemble_sum.to_netcdf(f"{MORTALITY_DIR}Mortality_global_sum_CESM2_{scenario}_{dates}.nc")
```
