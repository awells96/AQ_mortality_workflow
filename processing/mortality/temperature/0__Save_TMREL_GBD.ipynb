# Save TMREL data as netCDF files

EDIT **** The GBD RR curves can be downloaded [here](https://doi.org/10.6069/vkdr-qy60) in csv format. This script converts them to netCDF.

```python
import os
import re
import glob
import pandas as pd
import config
from utils.utils import require_dir
import pathlib
```

```python
# === Setup ===
DIR = require_dir(pathlib.Path(config.WORK_ROOT) / "GBD23" / "TMREL")

# Start with global TMREL - update to include each country
# "1" is global

in_file = os.path.join(DIR, "tmrel_1_summaries.csv")

df = pd.read_csv(in_file)

df = df[df["year_id"] == 2020].drop(columns=["year_id", "location_id"])
df = df.rename(columns={"meanTempCat": "annual_temperature"})

ds = df.set_index(["annual_temperature"]).to_xarray()

description = (
    "TMREL: Burkart K, Brauer M, Aravkin A et al. Estimating the "
    "cause-specific relative risks of non-optimal temperature "
    "on daily mortality: a two-part modelling approach applied "
    "to the Global Burden of Disease Study"
)
ds.attrs["description"] = description

out_file = "global_tmrel_summaries.nc"
out_path = os.path.join(DIR, out_file)
ds.to_netcdf(out_path)
print(f"Saved {out_path}")
```
