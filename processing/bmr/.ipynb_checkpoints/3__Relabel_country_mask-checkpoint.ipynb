```python
# Changing country labels for the GBD mask
```

```python
import xarray as xr
```

```python
# === Path config ===
BMR_DIR = "/glade/work/awells/air_quality/BMR/"
MASKS_DIR = "/glade/work/awells/air_quality/BMR/masks/country/"
```

```python
bmr = xr.open_dataarray(f"{BMR_DIR}GBD_BMR_Country_COPD_1990-2009.nc")
masks = xr.open_dataarray(f"{MASKS_DIR}GBD_Country_Masks_0.10.nc")
```

```python
countries_1 = set(bmr["country"].values)  # Country labels we want
countries_2 = set(masks["country"].values)  # Country labels we need to change
```

```python
# Show list of country names we will change FROM
mismatched = countries_2 - countries_1
print("Mismatched country names:", sorted(mismatched))
```

```python
# Show list of country names we will change TO
mismatched = countries_1 - countries_2
print("Mismatched country names:", sorted(mismatched))
```

```python
# Create a mapping of country names to change FROM: TO
name_mapping = {
    'Bolivia': 'Plurinational State of Bolivia',
    'Brunei': 'Brunei Darussalam',
    'Cape Verde': 'Republic of Cabo Verde',
    "Cote d'Ivoire": "Republic of Côte d'Ivoire",
    'Iran': 'Islamic Republic of Iran',
    'Laos': "Lao People's Democratic Republic",
    'Macedonia': 'North Macedonia',
    'Moldova': 'Republic of Moldova',
    'Monaco': 'Principality of Monaco',
    'Nauru': 'Republic of Nauru',
    'Niue': 'Republic of Niue',
    'North Korea': "Democratic People's Republic of Korea",
    'Palau': 'Republic of Palau',
    'San Marino': 'Republic of San Marino',
    'South Korea': 'Republic of Korea',
    'Swaziland': 'Kingdom of Eswatini',
    'Syria': 'Syrian Arab Republic',
    'Taiwan': 'Taiwan (Province of China)',
    'Tanzania': 'United Republic of Tanzania',
    'The Bahamas': 'Commonwealth of the Bahamas',
    'The Gambia': 'Republic of the Gambia',
    'United Kingdom': 'United Kingdom of Great Britain and Northern Ireland',
    'United States': 'United States of America',
    'Venezuela': 'Bolivarian Republic of Venezuela',
    'Vietnam': 'Socialist Republic of Viet Nam',
    'Virgin Islands, U.S.': 'United States Virgin Islands',
}
```

```python
# Change "masks" country names
masks = masks.assign_coords(
    country=[name_mapping.get(c, c) for c in masks['country'].values]
)
```

```python
# Save output
masks.to_netcdf(f"{MASKS_DIR}GBD_Country_Masks_0.10_newlabels.nc")
```
