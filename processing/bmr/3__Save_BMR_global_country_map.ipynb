# Save a global country-level map of BMR

Using `create_global_country_map` to save a (lat, lon) gridded BMR product for each mortality outcome

```python
import os
import xarray as xr
import config
from utils.utils import require_dir
import pathlib
from utils.utils import create_global_country_map
```

```python
# === Path config ===
BMR_DIR = require_dir(pathlib.Path(config.WORK_ROOT) / "BMR")
MASKS_DIR = require_dir(pathlib.Path(config.WORK_ROOT) / "BMR" / "masks" / "country", "country masks")
```

```python
# === Health variables ===
health_vars = ["COPD", "DIABETES", "ISCHEMIC_HEART_DISEASE",
               "LOWER_RESPIRATORY_INFECTIONS", "LUNG_CANCER",
               "STROKE", "DEMENTIA"]
```

```python
# === Set GBD version ===
GBD_version = "GBD23"
```

```python
for health_VAR in health_vars:
    print(f"Processing {health_VAR}")
    bmr_file = f"{GBD_version}_BMR_Country_{health_VAR}_newlabels_1990-2009.nc"
    bmr_path = os.path.join(BMR_DIR, bmr_file)
    bmr_country = xr.open_dataarray(bmr_path).sel(quantile="mean").load()  # central estimate
    BMR = create_global_country_map(bmr_country, MASKS_DIR)

    out_file = f"{GBD_version}_BMR_Global_Country_Map_{health_VAR}_1990-2009.nc"
    out_path = os.path.join(BMR_DIR, out_file)
    print(f"Saving to {out_path}")
    BMR.to_netcdf(out_path)

print("All processing complete.")
```
