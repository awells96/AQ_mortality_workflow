```python
# === Latitude weighting sum ===
def weighted_sum(da):
    weights = np.cos(np.deg2rad(da.lat))
    weights.name = "weights"
    new_da = da.weighted(weights).sum(("lon", "lat"))
    return new_da
```

```python
tot_pop = []
years = range(2000, 2110, 10)

for year in range(2000, 2110, 10):
    print(f"Processing {year}")
    if year == 2000:
        pop = xr.open_dataset(f"{POP_DIR}baseYr_total_{year}.nc4")["Band1"]
    else:
        pop = xr.open_dataset(f"{POP_DIR}ssp2_total_{year}.nc4")["Band1"]

    tot_pop.append(pop)

population_time_series = xr.concat(tot_pop, xr.DataArray(years, dims="year", name="year"))
```

```python
new_year = range(2000, 2101)

new_pop = population_time_series.interp(year=new_year)
```

```python
weighted_sum(new_pop).plot()
plt.grid()
```
