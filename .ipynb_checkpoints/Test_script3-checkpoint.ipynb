```python
import os
import xarray as xr
import regionmask
import matplotlib.pyplot as plt
import numpy as np
```

```python
BMR_DIR = "/glade/work/awells/air_quality/BMR/"
```

```python
bmr = xr.open_dataarray(f"{BMR_DIR}GBD_Country_BMR_COPD_1990-2009.nc")
masks = xr.open_dataarray(f"{BMR_DIR}GBD_Country_Masks_0.10_newlabels.nc")
```

```python
# Create DataArray filled with NaNs
nan_array = xr.DataArray(
    np.full((len(masks.lat), len(masks.lon)), np.nan),
    coords={"lat": masks.lat.values, "lon": masks.lon.values},
    dims=["lat", "lon"]
)
```

```python
for i in range(204):
    mask = masks.isel(country=i)
    country = masks.isel(country=i)["country"]
    bmr_country = bmr.sel(country=country)[1] # 1 is median
    nan_array = nan_array.where(mask == 0, bmr_country)
```

```python
nan_array.plot()
```
