```python
import os
import xarray as xr
import regionmask
import matplotlib.pyplot as plt
import numpy as np
```

```python
BMR_DIR = "/glade/work/awells/air_quality/BMR/"
POP_DIR = "/glade/work/awells/air_quality/SSP_pop/SSP2/"
OSDMA8_DIR = "/glade/work/awells/air_quality/CESM/OSDMA8_BC/"
```

```python
bmr = xr.open_dataarray(f"{BMR_DIR}GBD_global_BMR_COPD_1990-2009.nc").mean("time")
pop_2050 = xr.open_dataset(f"{POP_DIR}ssp2_total_regrid_2050.nc")["Band1"]
test_2050 = xr.open_dataarray(f"{OSDMA8_DIR}OSDMA8_BC_CESM2_SSP245_02_2040-2059.nc")
```

```python
beta = 1.074
TMREL = 32.4
```

```python
O3_diff = test_2050 - TMREL
TMREL_O3 = xr.where(O3_diff > 0, O3_diff, 0)
```

```python
RR = np.exp(beta * TMREL_O3)
```

```python
AF = (RR - 1)/RR
```

```python
M = AF * bmr[1] * pop_2050
```

```python
M.plot(vmax=1e7)
```
